## Notebook 1: Data Loading & Cleaning

### Project Overview
**Objective:** Predict whether a company will go bankrupt based on its financial ratios.

**Dataset:** Bankruptcies.csv

**Purpose of this notebook:** Load the raw data, inspect its structure, check its quality, remove problematic columns, and save a clean dataset for Exploratory Data Analysis (EDA) and model development.

### Workflow

1. Import libraries
2. Load dataset
3. Create backup and inspect structure
4. Review dataset metadata

6. Data cleaning
7. Save cleaned dataset
8. Summary


### 1. Import Libraries

pandas is used for loading and manipulating tabular data, numpy for numerical checks (such as detecting infinite values), and pathlib for file paths that work on any operating system.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Show all columns when displaying wide tables
pd.set_option("display.max_columns", None)
```

### 2. Load Dataset

The raw CSV is loaded into a DataFrame. Column names in this file contain leading spaces (for example `' ROA(C) before interest...'`), so whitespace is stripped immediately after loading. This prevents lookup errors in later steps.

```python
# Load raw data
data = pd.read_csv("Bankruptcies.csv")

# Remove leading/trailing spaces from column names
data.columns = data.columns.str.strip()

data.head()
```

### 3. Data Backup & Information

A working copy (df) is created so the original data (data) always stays untouched. df.info() then shows the number of rows, columns, non-null counts, and data type of each column.

```python
df = data.copy()
df.info()
```

### 4. Dataset Metadata

Examine the dataset dimensions, column names, numerical columns, and data types.

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

all_columns = df.columns.to_list()
num_columns = df.select_dtypes(include="number").columns.tolist()
cat_columns = df.select_dtypes(exclude="number").columns.tolist()

print("\nAll Columns:\n", all_columns)
print("\nNumeric Columns:\n", num_columns)
print("\nCategorical Columns:\n", cat_columns)
print("\nData Types:\n", df.dtypes)
```

### 5. Data Quality Check

Check column-name consistency, missing values, duplicate records, infinite values, constant columns, and the distribution of the target variable.

```python
# Check missing values
print("Total missing values:", df.isnull().sum().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check infinite values
numeric_df = df.select_dtypes(include="number")

print(
    "Infinite values:",
    np.isinf(numeric_df.to_numpy()).sum()
)

# Identify constant columns
constant_columns = [
    col for col in df.columns
    if df[col].nunique(dropna=False) <= 1
]

print("\nConstant columns:", constant_columns)

# Check target distribution
print("\nTarget distribution:")
print(df["Bankrupt?"].value_counts().sort_index())

print("\nTarget distribution (%):")
print(
    (df["Bankrupt?"].value_counts(normalize=True) * 100)
    .round(2)
    .sort_index()
)
```

#### Observations

- The dataset has **no missing values**, **no duplicate rows**, and **no infinite values**.
- **Net Income Flag** is a constant column (same value in every row), so it adds no predictive information and will be removed.
- The target is **highly imbalanced**: only about **3.23% (220 companies)** are bankrupt, while about **96.77% (6,599 companies)** are not.
- **Implication for modeling:** a model that always predicts "not bankrupt" would already reach about 97% accuracy, so accuracy alone is misleading. Later stages should use stratified train/test splits, an imbalance-handling technique, and metrics such as Recall, Precision, F1-score and ROC-AUC.

### 6. Data Cleaning

Remove the identified constant feature and validate the cleaned dataset while preserving the original data.

```python
# Create a cleaned copy of the dataset
df_clean = df.copy()

# Remove constant columns
df_clean = df_clean.drop(columns=constant_columns)

# Verify the result
print("Original dataset shape:", data.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("Removed columns:", constant_columns)
```

### 7. Save Cleaned Dataset

Save the cleaned dataset as a CSV file for use in subsequent EDA and model development.

```python
# Define the output file path
clean_data_path = Path(
    r"D:\Internship1\cleaned_Bankruptcy_data.csv"
)

# Create the folder if it does not exist
clean_data_path.parent.mkdir(parents=True, exist_ok=True)

# Save the cleaned dataset
df_clean.to_csv(clean_data_path, index=False)

# Confirm successful saving
print("File saved successfully:", clean_data_path.exists())
print("Saved location:", clean_data_path)
```
